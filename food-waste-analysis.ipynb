```python
import matplotlib.pyplot as plt

# Reasons why vegetables are wasted
reasons = [
    "Spoil quickly",
    "Bought too much",
    "Not used in time",
    "Appearance standards",
    "Storage problems",
    "Leftovers"
]

# Illustrative percentage values
waste = [25, 22, 20, 15, 10, 8]

# Create bar chart
plt.figure(figsize=(10, 6))

plt.bar(reasons, waste)

plt.title("Why Are Vegetables Among the Most Wasted Foods in the UK?")
plt.xlabel("Main Reasons")
plt.ylabel("Estimated Share (%)")

plt.xticks(rotation=30, ha="right")
plt.tight_layout()

plt.show()

# Explanation
print("\nWhy are vegetables highly wasted?")
print("--------------------------------")
print("1. Vegetables can spoil quickly, especially fresh leafy vegetables.")
print("2. People sometimes buy more vegetables than they can use.")
print("3. Vegetables may be forgotten in the refrigerator and pass their best quality.")
print("4. Some vegetables are rejected because of their size, shape or appearance.")
print("5. Poor storage can cause vegetables to lose freshness quickly.")
print("6. Cooked vegetables may be left over and eventually thrown away.")
```

```python
import matplotlib.pyplot as plt

# Illustrative food waste data
countries = ["UK", "US"]

# Estimated food waste per person per year (kg)
food_waste = [79, 325]

# Create bar chart
plt.figure(figsize=(8, 6))

bars = plt.bar(countries, food_waste)

plt.title("Food Waste Comparison: UK vs US")
plt.xlabel("Country")
plt.ylabel("Food Waste per Person per Year (kg)")

# Add values above the bars
for bar, value in zip(bars, food_waste):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value + 5,
        f"{value} kg",
        ha="center"
    )

plt.tight_layout()
plt.show()

# Print comparison
print("Food Waste Comparison")
print("----------------------")
print(f"UK: {food_waste[0]} kg per person per year")
print(f"US: {food_waste[1]} kg per person per year")

difference = food_waste[1] - food_waste[0]

print(f"\nDifference: {difference} kg per person per year")
print("The US has a higher food-waste level per person in this example.")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# ---------------------------------------------------------
# FOOD WASTE: UK vs US HOUSEHOLDS
# ---------------------------------------------------------
# US data: EPA 2023
# UK data: WRAP 2022
#
# Important:
# The methodologies are not perfectly identical.
# UK = household food and drink waste.
# US = consumer-level waste from grocery + food service.
# ---------------------------------------------------------

categories = [
    "Fresh vegetables",
    "Fresh fruit",
    "Bakery / grains",
    "Milk",
    "Meat",
    "Other dairy",
    "Poultry",
    "Processed vegetables",
    "Processed fruit",
    "Fish & seafood",
    "Eggs"
]

# ---------------------------------------------------------
# US DATA
# EPA 2023
# Pounds/person/year converted to kg
# Then multiplied by 4 for a four-person household
# ---------------------------------------------------------

us_lb_per_person = [
    31.0,   # Fresh vegetables
    22.6,   # Fresh fruit
    31.0,   # Grain products
    26.6,   # Fluid milk
    25.6,   # Meat
    17.8,   # Other dairy
    13.9,   # Poultry
    16.0,   # Processed vegetables
    6.7,    # Processed fruit
    5.1,    # Fish and seafood
    6.8     # Eggs
]

# Convert pounds to kilograms
LB_TO_KG = 0.453592

us_kg = [
    value * LB_TO_KG * 4
    for value in us_lb_per_person
]

# ---------------------------------------------------------
# UK DATA
#
# WRAP reports:
# Total household food waste = approximately 210 kg/household
# Fresh vegetables & salads ≈ 27%
# Fresh fruit ≈ 8%
#
# We only plot categories where the available UK data
# supports a reasonable estimate.
# ---------------------------------------------------------

uk_kg = [
    210 * 0.27,  # Fresh vegetables
    210 * 0.08,  # Fresh fruit
    np.nan,      # Bakery / grains
    np.nan,      # Milk
    np.nan,      # Meat
    np.nan,      # Other dairy
    np.nan,      # Poultry
    np.nan,      # Processed vegetables
    np.nan,      # Processed fruit
    np.nan,      # Fish & seafood
    np.nan       # Eggs
]

# ---------------------------------------------------------
# CREATE CHART
# ---------------------------------------------------------

x = np.arange(len(categories))
width = 0.36

fig, ax = plt.subplots(figsize=(14, 8))

# UK bars
uk_bars = ax.bar(
    x - width / 2,
    uk_kg,
    width,
    label="UK household"
)

# US bars
us_bars = ax.bar(
    x + width / 2,
    us_kg,
    width,
    label="US household (4 people)"
)

# ---------------------------------------------------------
# LABELS
# ---------------------------------------------------------

ax.set_title(
    "Household Food Waste by Food Category: UK vs US",
    fontsize=18,
    fontweight="bold",
    pad=20
)

ax.set_ylabel(
    "Food waste (kg per household per year)",
    fontsize=12
)

ax.set_xlabel(
    "Food category",
    fontsize=12
)

ax.set_xticks(x)
ax.set_xticklabels(
    categories,
    rotation=40,
    ha="right"
)

ax.legend()

# Remove unnecessary borders
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# Add values to US bars
for bar, value in zip(us_bars, us_kg):
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        value + 2,
        f"{value:.1f}",
        ha="center",
        va="bottom",
        fontsize=9
    )

# Add values to available UK bars
for bar, value in zip(uk_bars, uk_kg):
    if not np.isnan(value):
        ax.text(
            bar.get_x() + bar.get_width() / 2,
            value + 2,
            f"{value:.1f}",
            ha="center",
            va="bottom",
            fontsize=9
        )

# Footnote
plt.figtext(
    0.5,
    0.01,
    "Sources: WRAP UK household food waste 2022; US EPA consumer food waste 2023. "
    "UK category estimates are approximate. US methodology is not directly equivalent.",
    ha="center",
    fontsize=9
)

plt.tight_layout(rect=[0, 0.04, 1, 1])

plt.show()
```
