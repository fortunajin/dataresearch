```python
import matplotlib.pyplot as plt
import numpy as np

# Approximate values per 100 g of avocado flesh.
# Nutrition values vary by variety and source.
nutrients = {
    "Energy (kcal)": 177,
    "Total fat (g)": 17.0,
    "Monounsaturated fat (g)": 10.0,
    "Fibre (g)": 3.1,
    "Protein (g)": 1.8,
    "Carbohydrate (g)": 1.7
}

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Chart 1: Nutritional profile
names = list(nutrients.keys())
values = list(nutrients.values())

bars = axes[0].barh(names, values)
axes[0].invert_yaxis()
axes[0].set_title("Avocado Nutritional Profile")
axes[0].set_xlabel("Amount per 100 g")
axes[0].grid(axis="x", alpha=0.25)

for bar, value in zip(bars, values):
    axes[0].text(
        value + max(values) * 0.015,
        bar.get_y() + bar.get_height() / 2,
        f"{value:g}",
        va="center"
    )

# Chart 2: Evidence-based verdict, not a medical score
countries = ["United Kingdom", "United States"]
verdict = [1, 1]

axes[1].bar(countries, verdict)
axes[1].set_ylim(0, 1.35)
axes[1].set_yticks([0, 1])
axes[1].set_yticklabels(["Not rated", "Nutrient-dense"])
axes[1].set_title("Is Avocado a Superfood?")
axes[1].text(
    0.5, 0.55,
    "YES, in the popular sense\n"
    "Both countries:\n"
    "• Healthy unsaturated fats\n"
    "• Fibre and potassium\n"
    "• Vitamins and folate\n\n"
    "But 'superfood' is not an\n"
    "official scientific category.",
    transform=axes[1].transAxes,
    ha="center",
    va="center",
    wrap=True
)
axes[1].set_ylabel("Nutrition assessment")
axes[1].grid(axis="y", alpha=0.25)

fig.suptitle(
    "Avocado: A Nutrient-Dense Food in the UK and US",
    fontsize=16,
    fontweight="bold"
)
plt.tight_layout()
plt.savefig("avocado_superfood_uk_us.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python

import matplotlib.pyplot as plt
import numpy as np

# ------------------------------------
# TOP 10 SUPERFOODS: UK AND US
# Scores are illustrative, not official
# scientific nutrition rankings.
# ------------------------------------

foods = [
    "Leafy greens",
    "Berries",
    "Oily fish",
    "Beans & lentils",
    "Nuts & seeds",
    "Whole grains",
    "Cruciferous vegetables",
    "Plain yogurt",
    "Avocado",
    "Eggs"
]

scores = [9.7, 9.5, 9.5, 9.4, 9.3, 9.1, 9.0, 8.8, 8.7, 8.5]

benefits = [
    "Vitamin K, folate",
    "Fibre, polyphenols",
    "Omega-3, protein",
    "Plant protein, fibre",
    "Healthy fats, minerals",
    "Fibre, whole-grain nutrients",
    "Fibre, vitamin C",
    "Protein, calcium",
    "Unsaturated fats, fibre",
    "Protein, choline, B12"
]

colors = [
    "#238F83", "#8064A2", "#3F78A8", "#D98B36",
    "#B77B45", "#C5A128", "#39966D", "#5D8F80",
    "#69AE91", "#D66A5D"
]

# Chart styling
plt.rcParams["font.family"] = "DejaVu Sans"

fig, ax = plt.subplots(figsize=(13, 8))
fig.patch.set_facecolor("#F7F8FA")
ax.set_facecolor("#F7F8FA")

y = np.arange(len(foods))

bars = ax.barh(
    y, scores,
    color=colors,
    height=0.65,
    edgecolor="white",
    linewidth=1.2
)

# Labels and layout
ax.set_yticks(y)
ax.set_yticklabels(foods, fontsize=11)
ax.invert_yaxis()
ax.set_xlim(0, 13)
ax.set_xticks(np.arange(0, 11, 2))
ax.set_xlabel(
    "Illustrative nutrient-diversity score (out of 10)",
    labelpad=12
)

ax.set_title(
    "TOP 10 SUPERFOODS",
    fontsize=22,
    fontweight="bold",
    loc="left",
    pad=45,
    color="#203040"
)

ax.text(
    0, 1.025,
    "UK & US | Nutrient-rich foods for everyday eating",
    transform=ax.transAxes,
    fontsize=11,
    color="#5D6D7E"
)

# Add score and nutrient benefit labels
for bar, score, benefit in zip(bars, scores, benefits):
    middle = bar.get_y() + bar.get_height() / 2

    ax.text(
        score + 0.12, middle,
        f"{score:.1f}",
        va="center",
        fontsize=10,
        fontweight="bold",
        color="#203040"
    )

    ax.text(
        10.35, middle,
        benefit,
        va="center",
        fontsize=9,
        color="#4E5D6C"
    )

# Clean professional appearance
ax.xaxis.grid(True, color="#DCE2E8", linewidth=0.8)
ax.set_axisbelow(True)

for spine in ax.spines.values():
    spine.set_visible(False)

ax.tick_params(axis="both", length=0, pad=8)

# Footer
fig.text(
    0.25, 0.075,
    "KEY BENEFITS: Fibre | Healthy fats | Protein | Vitamins & minerals",
    fontsize=10,
    fontweight="bold",
    color="#34495E"
)

fig.text(
    0.25, 0.04,
    "Scores are illustrative educational ratings, not official scientific measurements.",
    fontsize=8.5,
    color="#657786"
)

plt.subplots_adjust(
    left=0.25, right=0.96,
    top=0.82, bottom=0.16
)

# Save a high-resolution image
plt.savefig(
    "top10_superfoods_uk_us.png",
    dpi=300,
    bbox_inches="tight",
    facecolor=fig.get_facecolor()
)

plt.show()
```
