```python
import matplotlib.pyplot as plt
import numpy as np

# ============================================================
# LONDON vs CALIFORNIA JOB OPPORTUNITIES — OCTOBER 2026
# ============================================================
#
# Data sources:
# London: ONS / Greater London Authority
# California: California Employment Development Department
#
# IMPORTANT:
# October 2026 official data are not yet complete.
# Therefore, this chart uses the latest available data
# as of October 2026.
# ============================================================

locations = ["London", "California"]

# ------------------------------------------------------------
# 1. Unemployment rate
# ------------------------------------------------------------

unemployment = [
    6.8,   # London, 3 months ending July 2026
    5.1    # California, July 2026
]

# ------------------------------------------------------------
# 2. Employment / job growth
# ------------------------------------------------------------

job_growth = [
    -1.0,   # London: approximate payroll employment YoY direction
    0.6     # California: non-farm job growth YoY
]

# ------------------------------------------------------------
# 3. Create comparison chart
# ------------------------------------------------------------

x = np.arange(len(locations))
width = 0.35

fig, ax = plt.subplots(figsize=(10, 7))

bars1 = ax.bar(
    x - width / 2,
    unemployment,
    width,
    label="Unemployment rate (%)"
)

bars2 = ax.bar(
    x + width / 2,
    job_growth,
    width,
    label="Job growth (%)"
)

# ------------------------------------------------------------
# Chart title
# ------------------------------------------------------------

ax.set_title(
    "Job Market Comparison: London vs California",
    fontsize=18,
    fontweight="bold",
    pad=20
)

ax.set_ylabel(
    "Percentage (%)",
    fontsize=12
)

ax.set_xticks(x)
ax.set_xticklabels(locations, fontsize=12)

ax.axhline(
    0,
    linewidth=0.8
)

ax.legend()

# ------------------------------------------------------------
# Add numbers to bars
# ------------------------------------------------------------

def add_labels(bars):
    for bar in bars:
        height = bar.get_height()

        if height >= 0:
            position = height + 0.15
            va = "bottom"
        else:
            position = height - 0.15
            va = "top"

        ax.text(
            bar.get_x() + bar.get_width() / 2,
            position,
            f"{height:.1f}%",
            ha="center",
            va=va,
            fontsize=11,
            fontweight="bold"
        )

add_labels(bars1)
add_labels(bars2)

# ------------------------------------------------------------
# Professional formatting
# ------------------------------------------------------------

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()

plt.show()


# ============================================================
# PRINT SUMMARY
# ============================================================

print("\nJOB OPPORTUNITY COMPARISON")
print("=" * 40)

for i, location in enumerate(locations):

    print(f"\n{location}")
    print(f"Unemployment rate: {unemployment[i]:.1f}%")
    print(f"Job growth: {job_growth[i]:.1f}%")

print("\nInterpretation")
print("-" * 40)

if unemployment[0] > unemployment[1]:
    print(
        "London currently has a higher unemployment rate "
        "than California."
    )

if job_growth[1] > job_growth[0]:
    print(
        "California currently shows stronger employment growth "
        "than London."
    )
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# ============================================================
# STANFORD JOB APPLICATION & SUCCESS RATE TRACKER
# ============================================================

# Enter YOUR applications here
#
# Status options:
# Applied
# Rejected
# Interview
# Second Interview
# Offer
# Withdrawn

applications = [
    # Example data -- replace with your real applications

    {
        "Job": "Administrative Associate",
        "Category": "Administration",
        "Date": "2026-10-01",
        "Status": "Applied"
    },

    {
        "Job": "Finance Specialist",
        "Category": "Finance",
        "Date": "2026-10-03",
        "Status": "Applied"
    },

    {
        "Job": "Student Services Administrator",
        "Category": "Student Services",
        "Date": "2026-10-05",
        "Status": "Interview"
    },

    {
        "Job": "Restaurant / Food Services Manager",
        "Category": "Food Services",
        "Date": "2026-10-06",
        "Status": "Rejected"
    }
]

df = pd.DataFrame(applications)

df["Date"] = pd.to_datetime(df["Date"])

# ============================================================
# BASIC APPLICATION STATISTICS
# ============================================================

total_applications = len(df)

interviews = df[
    df["Status"].isin(["Interview", "Second Interview", "Offer"])
].shape[0]

offers = df[
    df["Status"] == "Offer"
].shape[0]

rejections = df[
    df["Status"] == "Rejected"
].shape[0]

# ============================================================
# SUCCESS RATES
# ============================================================

if total_applications > 0:
    interview_rate = interviews / total_applications * 100
else:
    interview_rate = 0

if total_applications > 0:
    offer_rate = offers / total_applications * 100
else:
    offer_rate = 0

if interviews > 0:
    interview_to_offer_rate = offers / interviews * 100
else:
    interview_to_offer_rate = 0


print("\n========================================")
print("STANFORD JOB APPLICATION ANALYSIS")
print("========================================")

print(f"Total applications:        {total_applications}")
print(f"Interviews:                {interviews}")
print(f"Offers:                    {offers}")
print(f"Rejections:                {rejections}")

print("\n----------------------------------------")
print("SUCCESS RATES")
print("----------------------------------------")

print(f"Application → Interview:   {interview_rate:.1f}%")
print(f"Application → Offer:       {offer_rate:.1f}%")
print(f"Interview → Offer:         {interview_to_offer_rate:.1f}%")


# ============================================================
# SUCCESS RATE BY JOB CATEGORY
# ============================================================

category_results = []

for category, group in df.groupby("Category"):

    applications_count = len(group)

    interviews_count = group[
        group["Status"].isin(
            ["Interview", "Second Interview", "Offer"]
        )
    ].shape[0]

    offers_count = group[
        group["Status"] == "Offer"
    ].shape[0]

    interview_rate = (
        interviews_count / applications_count * 100
    )

    offer_rate = (
        offers_count / applications_count * 100
    )

    category_results.append({
        "Category": category,
        "Applications": applications_count,
        "Interviews": interviews_count,
        "Offers": offers_count,
        "Interview Rate %": interview_rate,
        "Offer Rate %": offer_rate
    })

category_df = pd.DataFrame(category_results)

print("\n========================================")
print("SUCCESS BY JOB CATEGORY")
print("========================================")

print(
    category_df.to_string(index=False)
)


# ============================================================
# CHART 1 — APPLICATIONS BY CATEGORY
# ============================================================

plt.figure(figsize=(10, 6))

category_counts = df["Category"].value_counts()

plt.bar(
    category_counts.index,
    category_counts.values
)

plt.title(
    "Stanford Job Applications by Category",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Job Category")
plt.ylabel("Number of Applications")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


# ============================================================
# CHART 2 — INTERVIEW SUCCESS RATE
# ============================================================

plt.figure(figsize=(10, 6))

plt.bar(
    category_df["Category"],
    category_df["Interview Rate %"]
)

plt.title(
    "Stanford Application → Interview Rate",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Job Category")
plt.ylabel("Interview Rate (%)")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


# ============================================================
# CHART 3 — OFFER SUCCESS RATE
# ============================================================

plt.figure(figsize=(10, 6))

plt.bar(
    category_df["Category"],
    category_df["Offer Rate %"]
)

plt.title(
    "Stanford Application → Offer Rate",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Job Category")
plt.ylabel("Offer Rate (%)")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


# ============================================================
# RECOMMENDATION
# ============================================================

if len(category_df) > 0:

    best_category = category_df.loc[
        category_df["Interview Rate %"].idxmax(),
        "Category"
    ]

    print("\n========================================")
    print("RECOMMENDATION")
    print("========================================")

    print(
        f"Your strongest category based on interview rate: "
        f"{best_category}"
    )

print("\nKeep applying to multiple suitable Stanford positions.")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ============================================================
# ESTIMATED LIFETIME EARNINGS
# Education + Reading/Literacy Ability
# ============================================================

# IMPORTANT:
# These are ESTIMATES for visualization.
# They are NOT official lifetime-income statistics.
#
# Education earnings are anchored to recent U.S. BLS data.
# Literacy adjustments are illustrative and should not be
# interpreted as causal estimates.

# ------------------------------------------------------------
# ASSUMPTIONS
# ------------------------------------------------------------

working_years = 40

# Annual earnings are estimated from recent BLS
# median weekly earnings.
#
# Weekly earnings:
# Less than HS       $770
# High school        $966
# Some college       $1,097
# Bachelor's+        $1,740

education = {
    "Less than High School": 770 * 52,
    "High School": 966 * 52,
    "Some College / Associate": 1097 * 52,
    "Bachelor's+": 1740 * 52
}

# ------------------------------------------------------------
# READING / LITERACY ADJUSTMENTS
# ------------------------------------------------------------
#
# Illustrative adjustment:
#
# Low literacy       = -15%
# Medium literacy    = baseline
# High literacy      = +15%
#
# These percentages are NOT official OECD coefficients.
# They are included to demonstrate how literacy could be
# incorporated into a lifetime-income model.

literacy_adjustment = {
    "Low Reading Ability": 0.85,
    "Medium Reading Ability": 1.00,
    "High Reading Ability": 1.15
}

# ------------------------------------------------------------
# CALCULATE ESTIMATED LIFETIME INCOME
# ------------------------------------------------------------

rows = []

for education_level, annual_income in education.items():

    for reading_level, adjustment in literacy_adjustment.items():

        adjusted_annual_income = annual_income * adjustment

        lifetime_income = (
            adjusted_annual_income * working_years
        )

        rows.append({
            "Education": education_level,
            "Reading Ability": reading_level,
            "Annual Income": adjusted_annual_income,
            "Lifetime Income": lifetime_income
        })

df = pd.DataFrame(rows)

# ------------------------------------------------------------
# FORMAT NUMBERS
# ------------------------------------------------------------

df["Annual Income"] = df["Annual Income"].round(0)

df["Lifetime Income"] = df["Lifetime Income"].round(0)

print("\n==============================================")
print("ESTIMATED LIFETIME INCOME")
print("==============================================")

print(
    df.to_string(index=False)
)


# ============================================================
# CHART 1
# Lifetime income by education and reading ability
# ============================================================

pivot = df.pivot(
    index="Education",
    columns="Reading Ability",
    values="Lifetime Income"
)

ax = pivot.plot(
    kind="bar",
    figsize=(12, 7)
)

plt.title(
    "Estimated Lifetime Income by Education and Reading Ability",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Education Level")

plt.ylabel(
    "Estimated Lifetime Income (USD)"
)

plt.xticks(rotation=30)

plt.legend(
    title="Reading Ability"
)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.show()


# ============================================================
# CHART 2
# Effect of reading ability within each education level
# ============================================================

plt.figure(figsize=(12, 7))

for reading_level in literacy_adjustment.keys():

    subset = df[
        df["Reading Ability"] == reading_level
    ]

    plt.plot(
        subset["Education"],
        subset["Lifetime Income"],
        marker="o",
        linewidth=2,
        label=reading_level
    )

plt.title(
    "Lifetime Income: Education vs Reading Ability",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Education Level")

plt.ylabel(
    "Estimated Lifetime Income (USD)"
)

plt.xticks(rotation=30)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.show()


# ============================================================
# FIND HIGHEST AND LOWEST ESTIMATES
# ============================================================

highest = df.loc[
    df["Lifetime Income"].idxmax()
]

lowest = df.loc[
    df["Lifetime Income"].idxmin()
]

print("\n==============================================")
print("HIGHEST ESTIMATED LIFETIME INCOME")
print("==============================================")

print(highest)


print("\n==============================================")
print("LOWEST ESTIMATED LIFETIME INCOME")
print("==============================================")

print(lowest)


# ============================================================
# DIFFERENCE BETWEEN HIGH AND LOW
# ============================================================

difference = (
    highest["Lifetime Income"]
    - lowest["Lifetime Income"]
)

print("\nEstimated lifetime-income difference:")
print(f"${difference:,.0f}")
```
