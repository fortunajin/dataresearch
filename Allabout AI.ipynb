```python

import matplotlib.pyplot as plt
import numpy as np

# Popular AI products and their main uses
products = [
    "ChatGPT",
    "Google Gemini",
    "Claude",
    "Microsoft Copilot",
    "Perplexity",
    "Canva AI",
    "Midjourney",
    "Runway",
    "GitHub Copilot",
    "ElevenLabs"
]

uses = [
    "Writing & chat",
    "Research",
    "Office work",
    "Coding",
    "Image design",
    "Video & audio"
]

# Illustrative scores: 1 = limited fit, 5 = strong fit
scores = np.array([
    [5, 4, 3, 4, 4, 3],
    [5, 4, 4, 3, 4, 3],
    [5, 4, 3, 5, 2, 2],
    [4, 3, 5, 3, 3, 2],
    [3, 5, 2, 2, 1, 1],
    [3, 2, 4, 1, 5, 3],
    [1, 2, 1, 1, 5, 2],
    [2, 2, 2, 1, 3, 5],
    [2, 2, 3, 5, 1, 1],
    [2, 1, 2, 1, 2, 5]
])

fig, ax = plt.subplots(figsize=(13, 8))
fig.patch.set_facecolor("#F7F8FA")
ax.set_facecolor("#F7F8FA")

# Colourful heatmap
image = ax.imshow(
    scores,
    cmap="YlGnBu",
    vmin=1,
    vmax=5,
    aspect="auto"
)

ax.set_xticks(range(len(uses)))
ax.set_xticklabels(uses, rotation=25, ha="right", fontsize=10)

ax.set_yticks(range(len(products)))
ax.set_yticklabels(products, fontsize=11)

# Display scores inside every cell
for row in range(len(products)):
    for col in range(len(uses)):
        score = scores[row, col]
        ax.text(
            col, row, str(score),
            ha="center",
            va="center",
            fontsize=10,
            fontweight="bold",
            color="white" if score >= 4 else "#263238"
        )

ax.set_title(
    "AI PRODUCTS & THEIR MAIN USES",
    loc="left",
    fontsize=19,
    fontweight="bold",
    color="#203040",
    pad=24
)

ax.text(
    0, 1.025,
    "A practical comparison of 10 popular AI tools",
    transform=ax.transAxes,
    fontsize=11,
    color="#5D6D7E"
)

colourbar = fig.colorbar(image, ax=ax, pad=0.025)
colourbar.set_label("Qualitative fit (1–5)")
colourbar.set_ticks([1, 2, 3, 4, 5])

for spine in ax.spines.values():
    spine.set_visible(False)

ax.tick_params(length=0, pad=8)

fig.text(
    0.10, 0.035,
    "Note: Scores are illustrative guidance, not actual user numbers or market share.",
    fontsize=9,
    color="#657786"
)

plt.tight_layout(rect=[0, 0.07, 1, 1])

# Save chart as a high-resolution image
plt.savefig(
    "ai_products_use_cases_chart.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
